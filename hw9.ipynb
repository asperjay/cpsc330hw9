The COSMOS 2020 study surveys over 1.7 million photometric sources over a small section of sky. Most of the galaxies are normal and uninteresting amongst millions of other stars, but in other photometric surveys we find unusual galaxies, such as the Mice Galaxies - two galaxies slowly but violently colliding, throwing out long tidal tails of stars and gas hundreds of thousands of light-years long - or Hoag's object - a ring of young blue stars surrounding a center of much older yellow stars, with a uniform gap of about 50,000 light years in between, a particularly uncommon shape. These 'peculiar' galaxies that cannot be very convincingly placed into traditional typologies can be uncovered by new photometric surveys and are often at the forefront of scientific progress in this field; they give insights into star formation (why does Hoag's object have a donut shape instead of a more traditional spiral shape?) and gravitational effects (what does the shape of the tidal tails of the Mice galaxies tell us about dark matter?).\
